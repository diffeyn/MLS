```python
import pandas as pd
import os
import re
from datetime import datetime

this_year = datetime.now().year
last_year = this_year - 1

dfs26 = []
dfs25 = []

pattern = re.compile(r"player_stats_.*\.csv")

for file in files:
    if pattern.search(file):
        mod_time = datetime.fromtimestamp(os.path.getmtime(file)).year
        
        try:
            if mod_time == this_year:
                dfs26.append(pd.read_csv(file, dtype={'match_id': str}))
            elif mod_time == last_year:
                dfs25.append(pd.read_csv(file, dtype={'match_id': str}))
        except Exception as e:
            print(f"Error reading {file}: {e}")

# Safety check before concat
df26 = pd.concat(dfs26, ignore_index=True) if dfs26 else pd.DataFrame()
df25 = pd.concat(dfs25, ignore_index=True) if dfs25 else pd.DataFrame()

print(f"2026: {len(df26)} rows across {len(dfs26)} files")
print(f"2025: {len(df25)} rows across {len(dfs25)} files")
```

```python
month_map = {
    "january": "01", "february": "02", "march": "03", "april": "04",
    "may": "05", "june": "06", "july": "07", "august": "08",
    "september": "09", "october": "10", "november": "11", "december": "12"
}

days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

# Step 1: Isolate first line immediately (handles \n, \r\n, \r)
df26['date'] = df26['date'].apply(lambda x: x.splitlines()[0].strip())

# Step 2: Remove day names
df26['date'] = df26['date'].apply(
    lambda x: ' '.join([word for word in x.split() if word not in days])
).str.strip()

# Step 3: Lowercase
df26['date'] = df26['date'].str.lower()

# Step 4: Replace month name with number
df26['date'] = df26['date'].apply(
    lambda x: re.sub(
        r"([A-Za-z]+)",
        lambda m: month_map[m.group(1).lower()],
        x
    )
)

# Step 5: Append year if missing
df26['date'] = df26['date'].apply(
    lambda x: x + " 2026" if re.search(r"\d{4}$", x) is None else x
)

df26['date'] = pd.to_datetime(df26['date'], format="%m %d %Y")

df26
```

```python
def norm_team(s):
    if pd.isna(s):
        return ""
    return (
        str(s).lower()
        .strip()
        .replace(".", "")
        .replace("’", "'")
    )

# 1) Build a normalized map from your official team_map
team_map_norm = {norm_team(k): v for k, v in team_map.items()}

# 2) Add aliases that appear in your current data
aliases = {
    "chicago fire": "CHI",                 # vs Chicago Fire FC
    "houston dynamo": "HOU",               # vs Houston Dynamo FC
    "dc united": "DC",                     # vs D.C. United
    "los angeles fc": "LAFC",              # vs Los Angeles Football Club
    "orlando city sc": "ORL",              # vs Orlando City
    "st louis city sc": "STL",             # vs St. Louis CITY SC
    "inter miami": "MIA",                  # vs Inter Miami CF
    "minnesota united fc": "MIN",          # already there but keep explicit
}
abbrs = set(team_map_norm.values())

df["club_norm"] = df["club"].apply(norm_team)

df["club_abbr"] = df["club_norm"].apply(
    lambda x: x.upper() if x.upper() in abbrs else team_map_norm.get(x)
)
```

```python
def clean_players(df):
    df = df.rename(columns={
        'Player': 'player_name',
        'Mins': 'minutes',
        'G' : 'goals',
        'xG': 'expected_goals',
        'Conv%' : 'shot_conv_perc',
        'SOT' : 'on_target',
        'Pass%' : 'pass_perc',
        'A' : 'assists',
        'P' : 'passes',
        'Cross' : 'cross',
        'CK' : 'corner_kick',
        'KP' : 'key_pass',
        'AD' : 'aerial',
        'AD%' : 'aerial_perc',
        'FC' : 'fouls',
        'FS' : 'fouls_against',
        'OFF' : 'offside',
        'YC' : 'yellow_card',
        'RC' : 'red_card'
    })
    
    df = df[['match_id'] + [col for col in df.columns if col != 'match_id']]
    
    return df
```
