```python
from sqlalchemy import Table, MetaData, insert

metadata = MetaData()
metadata.reflect(bind=engine)

df = df.astype(object).where(pd.notna(df), None)


# Access any table like a dictionary
teams_table = metadata.tables['match_team_stats']

stmt = insert(teams_table).prefix_with("IGNORE").values(df.to_dict(orient='records'))
```

```python
with engine.connect() as conn:
    conn.execute(stmt)
    conn.commit()
```
